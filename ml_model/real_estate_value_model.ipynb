```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('/home/timur/HM_course/HM3/all_v2.csv')
```

```python
data.info()
```

```python
data.drop(['building_type', 'date', 'time', 'geo_lon', 'geo_lat', 'object_type'], axis= 1 , inplace= True )
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split, GridSearchCV
```

```python
data = data.loc[data['price'] > 100000]
data = data.loc[data['region'] <= 85]
data = data.loc[data['price'] < 1000000000]
data = data.sample(frac=0.25)
```

```python
data['price'].count()
```

```python
X = data.drop('price', axis=1)
y = data['price']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

```python
rf = RandomForestRegressor()
```

```python
rf.fit(X, y)
```

```python
rf.score(X_test, y_test)
```

```python
rf.predict(np.array([77, 15, 15, 4, 130, 20]).reshape(1, -1))
```
